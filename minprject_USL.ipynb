```python
# load and Understand data

data= {
    "StudyHours":[2,5,7,4,6,2,6,10,11,9,2,1,8,3,5],
    "Atendence": [60,80,40,53,90,78,60,50,68,79,99,98,78,89,80],
    "Pastscore": [78,56,67,45,65,64,79,89,64,93,90,98,60,94,96],
    "Internet": ["Yes","Yes","No","Yes","No","No","Yes","Yes","No","Yes","No","Yes","Yes","Yes","No"],
    "Sleephours": [4,6,2,8,4,7,5,9,3,7,2,6,7,7,3],
    "Passed" :  ["No","Yes","Yes","No","No","No","Yes","Yes","No","Yes","No","Yes","Yes","Yes","No"]
}
df=pd.DataFrame(data)
print(df.head())
print(f"Rows: {df.shape[0]},column: {df.shape[1]}")
```

###  Preprocessing (Clean Data)
   - If we want that convert the c atagoricall data into numerical data then we use to to method called "LabelEncoder" or " GetDummies"

        - LabelEncoder : Use when we have only two categories in the given columns like "Yes" or "No"
        - Getdummies : Use when the given columns have multiple cetogories in the columns like more then tewo cetogories


```python
le=LabelEncoder()

df["Internet"]=le.fit_transform(df["Internet"])
df["Passed"]=le.fit_transform(df["Passed"])

print("/n Label Encoded Data")
print(df.head())

print("/n types")
print(df.dtypes)


plt.figure(figsize=[8,6])

plt.hist(df["Internet"],df["Passed"],color='black',bins='True')
plt.title("Encoded Data(Visualization)")
plt.xlabel("Internet")
plt.ylable('Passed')
plt.show ()
```
